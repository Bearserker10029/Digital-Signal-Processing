```python
import numpy as np
import matplotlib.pyplot as plt

np.__version__
```

Crear una señal

```python
# Parametros de la senal sinusoidal
A = 20.0       # amplitud
f = 5         # frecuencia [Hz]
fs = 1000     # frecuencia de muestreo [Hz]
dur = 1.0     # duracion [s]
w= 2 * np.pi * f

# Eje de tiempo y senal
t = np.arange(0, dur, 1/fs) # crear vector de tiempo desde 0 hasta dur con paso de 1/fs
x = A * np.cos(w * t)*np.sin(w * t)

# Grafica
plt.figure(figsize=(10, 3))
plt.plot(t, x)
plt.title("Senal sinusoidal")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Multiplicacion por un tren de impulsos de Dirac (ADC ideal)
f_imp = 50  # frecuencia del tren de impulsos [Hz]
Ts = 1 / f_imp

Ns = int(round(fs / f_imp))  # muestras entre impulsos
p = np.zeros_like(x)
idx = np.arange(0, len(x), Ns)
p[idx] = 1.0

# Salida del muestreador ideal: y(t) = x(t) * p(t)
y_out = x * p

# Graficas
plt.figure(figsize=(12, 6))

plt.subplot(3, 1, 1)
plt.plot(t, x, color="tab:blue")
plt.title("x(t)")
plt.ylabel("Amplitud")
plt.grid(True, alpha=0.3)

plt.subplot(3, 1, 2)
plt.stem(t, p, linefmt="tab:orange", markerfmt=" ", basefmt="k-")
plt.title("p(t): tren de impulsos")
plt.ylabel("Impulso")
plt.grid(True, alpha=0.3)

plt.subplot(3, 1, 3)
plt.stem(t, y_out, linefmt="tab:green", markerfmt=" ", basefmt="k-")
plt.title("y(t) = x(t) * p(t)")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
