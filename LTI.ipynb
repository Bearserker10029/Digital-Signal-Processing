```python
import matplotlib.pyplot as plt
import numpy as np
from math import *
np.__version__
```

image

```python
w=np.linspace(-pi,pi,1000)

def H_freq(w):
    """Respuesta en frecuencia del sistema"""
    fase = -6*w - (2/np.pi)*w**2
    return np.exp(1j * fase)

H=H_freq(w)

magnitud=abs(H)
fase= np.unwrap(np.angle(H))
retardo=-np.diff(fase)/np.diff(w)

plt.figure(figsize=(12,8))
plt.subplot(3,1,1)
plt.plot(w,magnitud)
plt.title('Magnitud de H(w)')
plt.xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi], 
           ['$-\\pi$', '$-\\pi/2$', '0', '$\\pi/2$', '$\\pi$'])
plt.grid(True)

plt.subplot(3,1,2)
plt.plot(w,fase)
plt.title('Fase de H(w)')
plt.xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi], 
           ['$-\\pi$', '$-\\pi/2$', '0', '$\\pi/2$', '$\\pi$'])
plt.grid(True)

plt.subplot(3,1,3)
plt.plot(w[:-1],retardo)
plt.title('Retardo de H(w)')
plt.xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi], 
           ['$-\\pi$', '$-\\pi/2$', '0', '$\\pi/2$', '$\\pi$'])
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
n=np.arange(128)
w1= 0.2*pi
w2= 0.6*pi
x1=np.cos(w1*n)
x2=0.6*np.cos(w2*n)
x= x1 + x2

H1=H_freq(w1)
H2=H_freq(w2)

desfase1= np.angle(H1)
desfase2= np.angle(H2)

y= np.cos(w1*n + desfase1) + 0.6*np.cos(w2*n + desfase2)
```

```python
plt.figure(figsize=(12,6))
plt.plot(n,x,'m-', linewidth=2, label='Entrada x[n]', alpha=0.7)
plt.grid(True, alpha=0.3)
plt.plot(n,y,'c-', linewidth=2, label='Salida y[n]', alpha=0.7)
plt.grid(True, alpha=0.3)
plt.legend( loc='upper right', fontsize=12)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

Se evidencia que el sistema no introduce una distorsión de magnitud, sino de fase. Esto es debido a que la fase no es lineal y la magnitud es constante. La distorsión de fase puede causar que las señales se retrasen o adelanten en el tiempo, lo que puede afectar la calidad de la señal transmitida. Es importante tener en cuenta esta distorsión al diseñar sistemas de comunicación para garantizar una transmisión de señal clara y precisa, ya que la señal de entrada tendrá una forma diferente a la señal de salida aunque las amplitudes sean las mismas.
