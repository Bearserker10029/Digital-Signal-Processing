```python
import numpy as np
import matplotlib.pyplot as plt
from math import *
np.__version__
```

```python
a=0.8
n=np.linspace(0, 200, 201)
print(n)
```

```python
x=a**n
E= np.sum(np.abs(x)**2)
print(E)
Eaprox=1/(1-a**2)
print(Eaprox)
```

```python
w=np.linspace(-pi, pi, 1000)
X=1/(1-a*np.exp(-1j*w))
Enum=(1/(2*pi))*np.trapezoid(np.abs(X)**2,w)
print(Enum)
```

```python
err_t = (E - Eaprox)/Eaprox
err_f = (Enum - Eaprox)/Eaprox

print("E =", E)
print("Eaprox   =", Eaprox)
print("Enum   =", Enum)
print("err_t    =", err_t)
print("err_f    =", err_f)


plt.stem(n, x)
plt.xlabel("n"); plt.ylabel("x[n]"); plt.title("x[n] = (0.8)^n u[n]")
plt.show()
```
